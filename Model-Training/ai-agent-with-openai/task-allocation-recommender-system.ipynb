# Task Allocation Recommender System

## Data Model

- teamMemberID: A unique identifier for each team member.
- Name: The team member's full name.
- Skills: A comma-separated list of key skills.
- Experience: Total years of experience. 
- Availability: A percentage indicating current workload or availability.

```python
from langchain.chat_models import ChatOpenAI

llm = ChatOpenAI(openai_api_key="openai_api_key")

def create_prompt(task_details, team_members):
    # Part 1: Task Details
    task_prompt = f"Task Details:\nDescription: {task_details['description']}\nRequired Skills: {', '.join(task_details['skills'])}\nDeadline: {task_details['deadline']}\n\n"

    # Part 2: Team Members Details
    team_prompt = "Team Members:\n"
    for member in team_members:
        team_prompt += f"Name: {member['name']}, Skills: {', '.join(member['skills'])}, Experience: {member['experience']} years, Current Workload: {member['workload']}%\n"

    # Combine both parts
    full_prompt = task_prompt + team_prompt

    # Add instruction for the model
    full_prompt += "\nWhich team member is the best fit for this task, just give me his name, only the name?"

    return full_prompt

# Example usage
task_details = {
    "description": "Develop a new feature for the web application",
    "skills": ["JavaScript", "React"],
    "deadline": "2023-04-10"
}

team_members = [
    {"name": "Alice", "skills": ["JavaScript", "Python"], "experience": 5, "workload": 60},
    {"name": "Bob", "skills": ["Java", "React"], "experience": 3, "workload": 40},
    # Add more team members as needed
]

prompt = create_prompt(task_details, team_members)

# Use LangChain to process the prompt
response = llm.invoke(prompt)

# Print the response
print(response)
```
